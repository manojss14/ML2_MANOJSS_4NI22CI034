```python
# Find-S Algorithm

def find_s_algorithm(data, target_column):

    # Separate attributes and target
    attributes = data.drop(columns=[target_column])
    target = data[target_column]

    # Initialize hypothesis
    hypothesis = None

    # Process each training example
    for index, row in data.iterrows():

        # Check whether example is positive
        if str(row[target_column]).strip().lower() in ['yes', '1', 'true', 'positive']:

            example = row.drop(target_column).tolist()

            # First positive example
            if hypothesis is None:
                hypothesis = example.copy()

            else:
                # Compare attributes
                for i in range(len(hypothesis)):

                    if hypothesis[i] != example[i]:
                        hypothesis[i] = '?'

            print("Example:", example)
            print("Updated Hypothesis:", hypothesis)
            print("---------------------------")

    if hypothesis is None:
        print("No positive examples found!")
        return None

    return hypothesis
```

```python
# Specify target column name
target_column = df.columns[-1]

print("Target Column:", target_column)

# Run Find-S
final_hypothesis = find_s_algorithm(df, target_column)

print("\nFINAL HYPOTHESIS:")
print(final_hypothesis)
```

```python
if final_hypothesis is not None:

    print("\nMost Specific Hypothesis:")

    for column, value in zip(
        df.drop(columns=[target_column]).columns,
        final_hypothesis
    ):
        print(column, ":", value)
```
